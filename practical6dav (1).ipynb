```python
import pandas as pd

data = {
    "month": [
        "Jan", "Feb", "Mar", "Apr", "May", "Jun",
        "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
    ],
    "product": [
        "Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet",
        "Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet"
    ],
    "region": [
        "North", "South", "East", "West", "North", "South",
        "East", "West", "North", "South", "East", "West"
    ],
   "sales": [
    120000, 135000, 150000, 145000, 160000, 175000,
    180000, 170000, 190000, 210000, 230000, 250000
],
    "units_sold": [12,30,25,14,35,30,18,38,32,21,45,40
    ]
}

df = pd.DataFrame(data)

print(df)
```

Task 2 - create a line chart

```python
import matplotlib.pyplot as plt
plt.plot(df['month'], df['sales'], marker='o')
plt.title("Company Sales from January to December")
plt.xlabel("Month")
plt.ylabel("Sales")
```

task 3 - create a bar chart

```python
import matplotlib.pyplot as plt
total_sales = df.groupby("product")["sales"].sum()
print(total_sales)
total_sales.plot(kind="bar")

plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.show()
```

TASK 4 - CREATE A HISTOGRAM

```python
import matplotlib.pyplot as plt
plt.hist(df["sales"], )
plt.title("Distribution of Individual Sales Values")
plt.xlabel("Sales Value")
plt.ylabel("Number of Sales")
plt.show()
```

TASK 5 - CREATE A SCATTER PLOT

```python
import matplotlib.pyplot as plt
plt.scatter(df["units_sold"], df["sales"])
plt.title("Units Sold vs Total Sales")
plt.xlabel("Number of Units Sold")
plt.ylabel("Total Sales")

plt.show()
```

TASK 6 - CREATE A pie chart

```python
import matplotlib.pyplot as plt

sales = df.groupby('product')['sales'].sum()

plt.pie(sales, labels=sales.index, autopct='%1.1f%%')
plt.title('Share of Total Sales by Product')
plt.show()
```
